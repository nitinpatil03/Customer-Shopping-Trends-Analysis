```python
# ============================================================
# CUSTOMER BEHAVIOR ANALYSIS
# Python + MySQL
# ============================================================

# Step 1: Import required libraries
import pandas as pd
from sqlalchemy import create_engine


# ============================================================
# STEP 2: LOAD CSV FILE
# ============================================================

# CSV file is in the same folder as this notebook
df = pd.read_csv("customer_behavior.csv")

print("CSV loaded successfully!")
print(f"Number of rows: {len(df)}")
print(f"Number of columns: {len(df.columns)}")

# Display first 5 rows
display(df.head())


# ============================================================
# STEP 3: MYSQL DATABASE DETAILS
# ============================================================

username = "root"                  # MySQL username
password = "YOUR_MYSQL_PASSWORD"   # CHANGE THIS
host = "localhost"                 # MySQL server
port = "3306"                      # Default MySQL port
database = "customer_behavior"     # Your database name


# ============================================================
# STEP 4: CREATE MYSQL CONNECTION
# ============================================================

engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)

print("MySQL connection created successfully!")


# ============================================================
# STEP 5: LOAD DATA INTO MYSQL
# ============================================================

table_name = "customer"

df.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False
)

print(f"Data successfully loaded into MySQL table '{table_name}'!")


# ============================================================
# STEP 6: READ DATA BACK FROM MYSQL
# ============================================================

query = "SELECT * FROM customer LIMIT 5;"

result = pd.read_sql(query, engine)

print("Data retrieved successfully from MySQL:")

display(result)


# ============================================================
# STEP 7: CHECK NUMBER OF RECORDS
# ============================================================

count_query = "SELECT COUNT(*) AS total_records FROM customer;"

count_result = pd.read_sql(count_query, engine)

print("Total records in MySQL:")

display(count_result)
```
